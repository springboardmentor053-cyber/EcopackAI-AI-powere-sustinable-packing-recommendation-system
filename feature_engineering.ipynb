```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
engine = create_engine(
    "postgresql+psycopg2://postgres:1234@localhost:5432/ecopackai"
)
print("✅ Database connected")

```

```python
products = pd.read_sql("SELECT * FROM products", engine)
materials = pd.read_sql("SELECT * FROM materials", engine)

print(products.shape, materials.shape)
```

```python
fragility_map = {
    "low": 1,
    "medium": 2,
    "high": 3
}

products["fragility_score"] = products["fragility_level"].map(fragility_map)
```

```python
products["risk_score"] = (
    products["fragility_score"] +
    products["moisture_sensitivity"].map({"low":1, "medium":2, "high":3}) +
    products["temperature_sensitivity"].map({"low":1, "medium":2, "high":3})
)
```

```python
products["weight_log"] = np.log1p(products["weight_kg"])
products["shelf_life_months"] = products["shelf_life_days"] / 30
```

```python
materials["eco_score"] = (
    materials["biodegradability_score"] * 0.4 +
    materials["recyclability_percent"] * 0.4 -
    materials["co2_emission_score"] * 0.2
)
```

```python
materials["cost_category"] = pd.cut(
    materials["cost_per_unit"],
    bins=[0, 10, 30, 100],
    labels=["low", "medium", "high"]
)
```

```python
materials["strength_index"] = (
    materials["strength_mpa"] * materials["flexibility"].map({
        "low": 1,
        "medium": 1.5,
        "high": 2
    })
)
```

```python
products.to_csv("engineered_products.csv", index=False)
materials.to_csv("engineered_materials.csv", index=False)

print("✅ Feature engineering complete")
```
