MODULE 6 — Frontend UI Development (HTML/CSS/Bootstrap)

Goal

Build a Bootstrap UI that:

Collects product inputs

Calls Flask API

Displays AI recommendations

Shows ranking table + comparison metrics

1) UI Pages created
index.html (Input Form)

Contains form fields:

product_name (required)

product_category

product_weight_kg

fragility_level

temperature_sensitive

required_strength_score

preferred_biodegradability_score

max_packaging_cost_inr

User clicks:

Get Recommendations

results.html (Results Page)

Shows:

Product Summary (card)

Top Recommendation (card)

Material Ranking Table (with predictions + environment score)

Back button

2) UI uses Bootstrap + custom CSS

Files:

frontend/static/css/style.css

Bootstrap via CDN

Improvements added:

Better spacing and layout

Cards for summary + top recommendation

Table styling

Badges/chips for recyclability, bio, environment score

3) Frontend JS flow (script.js)
Step A: Save input on submit

form submit prevents page reload

saves productData into localStorage

redirects to results.html

Step B: On results page, call Flask API

Fetch call:

fetch("http://127.0.0.1:5000/api/recommend", {
  method: "POST",
  headers: { "Content-Type": "application/json" },
  body: JSON.stringify({ product_name: productData.product_name })
})

Step C: Fill table with API results

loops through data.recommendations

fills rows:
rank, material_name, pred_cost_inr, pred_co2_kg, recyclability, biodegradability, suitability_score, environment_score

Error handling:

if data.error → shows alert

4) Proof that UI is connected to backend

When Flask is running and I open UI:

results page shows real rankings

predictions & environment score match backend JSON

If backend not running:

UI shows “Failed to fetch recommendations from server”
