## Data Source

All dashboard insights are generated from:

- Recommendation history stored in PostgreSQL
- Material metadata
- Predicted cost & CO₂ values from ML models

The backend aggregates:
- Average cost by category
- Average CO₂ by category
- Material frequency (usage trends)
- Overall KPI metrics


```python
import pandas as pd
import psycopg2

# Example connection
conn = psycopg2.connect(
    host="localhost",
    database="ecopackai",
    user="postgres",
    password="your_password"
)

query = "SELECT * FROM recommendations"
df = pd.read_sql(query, conn)

df.head()
```

## KPI Metrics Implemented

- Total Recommendations
- Average Predicted Cost (INR)
- Average Predicted CO₂ (kg)


```python
total_runs = len(df)
avg_cost = df["pred_cost_inr"].mean()
avg_co2 = df["pred_co2_kg"].mean()

print("Total Runs:", total_runs)
print("Avg Cost:", round(avg_cost, 2))
print("Avg CO2:", round(avg_co2, 3))
```

We grouped by category to show:

- Average cost by category
- Average CO₂ by category
- Number of runs


```python
category_summary = df.groupby("category").agg(
    avg_cost=("pred_cost_inr", "mean"),
    avg_co2=("pred_co2_kg", "mean"),
    runs=("id", "count")
).reset_index()

category_summary
```

To show material trends, we counted frequency of recommended materials.


```python
material_counts = df["material_name"].value_counts().reset_index()
material_counts.columns = ["material_name", "count"]

material_counts.head(10)
```

Charts Implemented:
- Avg Cost by Category (Bar)
- Avg CO₂ by Category (Bar)
- Top Materials (Horizontal Bar)


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.bar(category_summary["category"], category_summary["avg_cost"])
plt.xticks(rotation=45)
plt.title("Average Cost by Category")
plt.ylabel("Cost (INR)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.bar(category_summary["category"], category_summary["avg_co2"])
plt.xticks(rotation=45)
plt.title("Average CO2 by Category")
plt.ylabel("CO2 (kg)")
plt.tight_layout()
plt.show()
```

```python
top10 = material_counts.head(10)

plt.figure(figsize=(8,5))
plt.barh(top10["material_name"], top10["count"])
plt.title("Top Recommended Materials")
plt.xlabel("Count")
plt.tight_layout()
plt.show()
```

On the Comparison page:

We implemented a Trade-off Scatter Chart using Chart.js.

X-axis: Predicted Cost  
Y-axis: Predicted CO₂  

This visually shows:
- Lower-left = best (low cost + low CO₂)
- Enables stakeholder decision making


Excel Export Implemented Using:
- openpyxl
- Structured table formatting
- KPI + Category + Material tables


```python
from openpyxl import Workbook

wb = Workbook()
ws = wb.active
ws.title = "Sustainability Report"

ws.append(["Total Runs", total_runs])
ws.append(["Average Cost", round(avg_cost,2)])
ws.append(["Average CO2", round(avg_co2,3)])

wb.save("sustainability_report.xlsx")
```
